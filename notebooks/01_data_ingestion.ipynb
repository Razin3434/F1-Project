```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")
PROC = Path("../data/processed")
PROC.mkdir(parents=True, exist_ok=True)
```

```python
# Load raw data
races = pd.read_csv(RAW / "races.csv")
results = pd.read_csv(RAW / "results.csv")
qualifying = pd.read_csv(RAW / "qualifying.csv")
lap_times = pd.read_csv(RAW / "lap_times.csv")
pit_stops = pd.read_csv(RAW / "pit_stops.csv")
drivers = pd.read_csv(RAW / "drivers.csv")
constructors = pd.read_csv(RAW / "constructors.csv")
driver_standings = pd.read_csv(RAW / "driver_standings.csv")
constructor_standings = pd.read_csv(RAW / "constructor_standings.csv")

# Print shape and preview
for name, df in [
    ("races", races),
    ("results", results),
    ("qualifying", qualifying),
    ("lap_times", lap_times),
    ("pit_stops", pit_stops),
    ("drivers", drivers),
    ("constructors", constructors),
    ("driver_standings", driver_standings),
    ("constructor_standings", constructor_standings)
]:
    print(f"{name}: {df.shape}")
    display(df.head())
```

```python
print("🚦 Missing value summary (top columns):")
for name, df in [
    ("races", races),
    ("results", results),
    ("qualifying", qualifying),
    ("lap_times", lap_times),
    ("pit_stops", pit_stops),
]:
    print(f"\n{name}")
    print(df.isnull().sum().sort_values(ascending=False).head(10))
```

```python
races.to_csv(PROC / "races.csv", index=False)
results.to_csv(PROC / "results.csv", index=False)
qualifying.to_csv(PROC / "qualifying.csv", index=False)
lap_times.to_csv(PROC / "lap_times.csv", index=False)
pit_stops.to_csv(PROC / "pit_stops.csv", index=False)
drivers.to_csv(PROC / "drivers.csv", index=False)
constructors.to_csv(PROC / "constructors.csv", index=False)
driver_standings.to_csv(PROC / "driver_standings.csv", index=False)
constructor_standings.to_csv(PROC / "constructor_standings.csv", index=False)
```
