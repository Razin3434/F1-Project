```python
import pandas as pd

circuits = pd.read_csv("../data/raw/circuits.csv")
print(circuits.head())
print(circuits.columns)
```

```python
circuits_clean = circuits[[
    'circuitId',
    'name',
    'country',
    'lat',
    'lng',
    'alt'
]].copy()

circuits_clean = circuits_clean.rename(columns={
    'name': 'circuit_name'
})
```

```python
street_circuits_refs = [
  'monaco',
    'marina_bay',     # Singapore
    'baku',
    'las_vegas',
    'jeddah',
    'miami',
    'albert_park',    # Melbourne
    'villeneuve' 
]

circuits_clean['circuitType'] = circuits['circuitRef'].apply(lambda x: 'street' if x in street_circuits_refs else 'permanent')
```

```python
circuits_clean['alt'] = circuits_clean['alt'].fillna(circuits_clean['alt'].median())

def altitude_bucket(alt):
    if alt < 100:
        return 'low'
    elif alt < 500:
        return 'medium'
    else:
        return 'high'

circuits_clean['altitude_bucket'] = circuits_clean['alt'].apply(altitude_bucket)
```

```python
print(circuits_clean.head())
print(circuits_clean['circuitType'].value_counts())
```

```python
circuits_clean.to_csv("../data/processed/circuits_clean.csv", index=False)
```
