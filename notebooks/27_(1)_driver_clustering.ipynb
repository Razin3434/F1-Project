```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/features_v2.csv")
df.head
```

```python
df["grid_delta"] = df["grid"] - df["positionOrder"]
```

```python
driver_stats = df.groupby("driverId").agg({
    "positionOrder": ["mean", "std"],
    "avgLapTime_s": ["mean", "std"],
    "grid_delta": "mean",
    "points": "mean"
}).reset_index()
```

```python
driver_stats.columns = [
    "driverId",
    "avg_finish",
    "finish_std",
    "avg_laptime",
    "laptime_std",
    "avg_grid_delta",
    "avg_points"
]
```

```python
driver_stats = driver_stats.dropna()

race_counts = df.groupby("driverId").size().reset_index(name="race_count")

driver_stats = driver_stats.merge(race_counts, on="driverId")

driver_stats = driver_stats[driver_stats["race_count"] >= 10]
```

```python
driver_stats.to_csv("../data/processed/driver_stats_v1.csv", index=False)
```
